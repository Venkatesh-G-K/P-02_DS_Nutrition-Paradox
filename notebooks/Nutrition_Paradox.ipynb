## Method-1_Function for data consolidating

```python
def get_data(url, age_label):
    response = requests.get(url).json()
    df = pd.DataFrame(response['value'])
    df['age_group'] = age_label
    return df
```

image.png

```python
df_ob_adults = get_data(urls['obesity_adults'], "adults")
df_ob_children = get_data(urls['obesity_children'], "children")

df_obesity = pd.concat([df_ob_adults, df_ob_children])

df_ma_adults = get_data(urls['malnutrition_adults'], "adults")
df_ma_children = get_data(urls['malnutrition_children'], "children")

df_malnutrition = pd.concat([df_ma_adults, df_ma_children])

```

```python
special_cases = {
    'GLOBAL': 'Global',
    'SEAR': 'South-East Asia Region',
    'WB_LMI': 'Low & Middle Income',
    'WB_HI': 'High Income',
    'WB_LI': 'Low Income',
    'EMR': 'Eastern Mediterranean Region',
    'EUR': 'Europe',
    'AFR': 'Africa',
    'WPR': 'Western Pacific Region',
    'AMR': 'Americas Region',
    'WB_UMI': 'Upper Middle Income'
}
```

```python
def convert_country(code):

    country = pycountry.countries.get(alpha_3=code)

    if country:
        return country.name

    return special_cases.get(code, code)
```

```python
df_obesity['Country'] = df_obesity['SpatialDim'].apply(convert_country)
df_malnutrition['Country'] = df_malnutrition['SpatialDim'].apply(convert_country)
```

## Obesity Level

```python
def obesity_level(x):

    if x >= 30:
        return "High"

    elif x >= 25:
        return "Moderate"

    return "Low"
```

```python
df_obesity['Obesity_Level'] = (
    df_obesity['Mean_Estimate'].apply(obesity_level)
)
```

## Malnutrition Level

```python
def malnutrition_level(x):

    if x >= 20:
        return "High"

    elif x >= 10:
        return "Moderate"

    return "Low"
```

```python
df_malnutrition['Malnutrition_Level'] = (
    df_malnutrition['Mean_Estimate'].apply(malnutrition_level)
)
```
